```python
# The location_zipcode column was originally stored as a numeric type, which resulted in the loss of leading zeros for valid ZIP codes 
# (e.g., 02128 becoming 2128). Since ZIP codes are identifiers rather than numeric quantities, the column was standardized to a string 
# data type to preserve formatting and allow for missing values.

# An attempt was made to recover missing ZIP codes by extracting them from the location field. However, analysis showed that 57,967 
# records do not contain any ZIP code information within the location string itself. As a result, ZIP values for these records cannot 
# be reliably inferred through string parsing alone.

# To maintain data integrity and avoid introducing assumptions, these values have been retained as missing (NA). Future enrichment 
# (e.g., ZIP inference using latitude/longitude) may be considered if higher geographic completeness is required.
```

```python
# Replacing NAN values in column location_street_name with pd.NA


# Note on location_street_name Data Handling
# The location_street_name column contains a limited number of missing values. An attempt was made to recover these 
# values by extracting the street name from the location field. However, further inspection showed that for every 
# record where location_street_name is missing, the corresponding location value is also missing or contains no usable address information.

# As a result, the street name cannot be reliably derived from other available columns. To preserve data integrity and avoid 
# introducing assumptions, these values have been retained as missing and standardized to NA.

df["location_street_name"] = df["location_street_name"].replace({np.nan:pd.NA})
```

```python
# Normalising the ward column and keeping the format "NumberNumber"
import re
word_to_remove = "Ward"
pattern = r'\b' + re.escape(word_to_remove) + r'\b'
df["ward"] = df['ward'].str.replace(pattern, '', regex=True, case=False).str.strip()

#Remove leading zeroes
df["ward"] = df["ward"].str.lstrip("0")

#Convert empty string creating by lstrip
df.loc[df["ward"].isin(["", "0"]), "ward"] = pd.NA

#Convert string to integer
df["ward"] = pd.to_numeric(df["ward"], errors="coerce").astype("Int64")

df["ward"].unique()

# When you first looked at the ward column, the issue was not missing data but inconsistent formatting. Some 
# records had “Ward 1”, some had “1”, some had “01”, and even values like “0” or “00”. Although some of these 
# represent the same district, grouping the data in its raw form would treat them as different categories. That 
# would split counts across multiple versions of the same ward and make your analysis inaccurate.

# What you did was standardize representation, not delete information. You removed the “Ward” prefix, stripped 
# leading zeros, converted invalid values like “0” into missing values, and then cast the column to a proper 
# numeric type. Now each ward is consistently stored as an integer between 1 and 22. This ensures correct grouping, 
# sorting, and aggregation in SQL, Excel, or BI tools.

# This is an important shift in thinking. Professional data work is less about filling nulls and more about 
# enforcing structural consistency. You did not change the meaning of the data. You made sure that the same 
# real-world district is represented in a single, consistent format across the dataset, which protects the 
# integrity of your future metrics and dashboards.
```
